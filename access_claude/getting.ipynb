```python
#create a api client
from anthropic import Anthropic
client =Anthropic()
model ="claude-haiku-4-5-20251001"
```

```python
#makeing a request 
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
```

```python
def add_user_message(user_message ,text ):
  user_message = {"role": "user", "content": text}
  messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
#these are helper messages 
```

```python
# Start with an empty message list
messages = []
# Add the initial user question of Define quantum computing in one sentence
add_user_message(messages, "Define quantum computing in one sentence")
# pass the list of message into "chat" to get the answer
answer = chat(messages)


# take the answer and add it as assistant message into our list
add_assistant_message(messages, answer)


# add in user's fowllo up qeustion

add_user_message(messages, "Write another sentence")

# Get the follow-up response with full context
final_answer = chat(messages)
final_answer
messages


```

```python
# make an inintial req
messages = []
#use a while true loop to run the chat bot forever
while True:
  #get user input
  user_input = input("> ")
  print(">" , user_input)
  # add user input to the list of messages
  add_user_message(messages, user_input)
  # call claude with chat function
  answer = chat(messages)
  add_assistant_message(messages, answer)
  # print the answer
  print("-------")
  print(answer)
  print("-------")
```
